# portfolio opimization automatically 

```python
from scipy.optimize import minimize
```

```python
def portfolio_variance(weights):
    return (
        weights.T
        @ annualized_covariance.values
        @ weights
    )
```

```python
weights = np.array([1/6]*6)

portfolio_variance(weights)
```

```python
initial_weights = np.array([1/6] * 6)
constraints = { 
    "type": "eq",
    "fun": lambda weights: np.sum(weights) - 1 # All sum of weights should be 1
}
bounds = [(0, 1)] * 6 # no short selling, each array can't be minus
```

```python
result = minimize(
    portfolio_variance,
    initial_weights,
    method="SLSQP", #Optimization Algorithm
    bounds=bounds,
    constraints=constraints
)
```

```python
print(result.message)
```

```python
optimal_weights = result.x

for ticker, weight in zip(close_prices.columns, optimal_weights):
    print(f"{ticker}: {weight:.2%}")
```

```python
optimized_return = np.dot(
    optimal_weights,
    annualized_returns
)


optimized_variance = (
    optimal_weights.T
    @ annualized_covariance.values
    @ optimal_weights
)

optimized_volatility = np.sqrt(optimized_variance)

optimized_sharpe = (
    optimized_return / optimized_volatility)
```

```python
print(f"Optimized Return: {optimized_return:.2%}")
print(f"Optimized Volatility: {optimized_volatility:.2%}")
print(f"Optimized Sharpe Ratio: {optimized_sharpe:.3f}")
```

# Out Of Sample Backtesting 

```python
split_index = int(len(close_prices) * 0.8)

train_prices = close_prices.iloc[:split_index]
test_prices = close_prices.iloc[split_index:]
```

```python
print("Train:")
print(train_prices.index.min())
print(train_prices.index.max())

print("\nTest:")
print(test_prices.index.min())
print(test_prices.index.max()) #train = 2018-2024, test = 2024-2026
```

```python
train_returns = train_prices.pct_change().dropna()
train_annualized_returns = train_returns.mean() * 252
train_covariance = train_returns.cov()
train_annualized_covariance = train_covariance * 252
```

```python
def train_portfolio_variance(weights):
    return (
        weights.T
        @ train_annualized_covariance.values
        @ weights
    )
    constraints = {
    "type": "eq",
    "fun": lambda weights: np.sum(weights) - 1
}
bounds = [(0, 1)] * len(close_prices.columns)
initial_weights = np.array(
    [1 / len(close_prices.columns)] * len(close_prices.columns)
)
train_minvar_result = minimize(
    train_portfolio_variance,
    initial_weights,
    method="SLSQP",
    bounds=bounds,
    constraints=constraints
)
train_minvar_weights = train_minvar_result.x

for ticker, weight in zip(
    close_prices.columns,
    train_minvar_weights
):
    print(f"{ticker}: {weight:.2%}") 
```

```python
test_returns = test_prices.pct_change().dropna()

test_portfolio_returns = np.dot(
    test_returns,
    train_minvar_weights
)
cumulative_test_returns = (
    1 + test_portfolio_returns
).cumprod() # use .cumulaitve product for applying interest
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    test_prices.index[1:],
    cumulative_test_returns
)

plt.xlabel("Date")
plt.ylabel("Growth of $1")
plt.title("Out-of-Sample Minimum Variance Portfolio")
plt.grid(True)

plt.show()
```

# Add funtion that calculate the transaction cost

```python
transaction_cost = 0.0025
```

```python
initial_weights = np.array(
    [1 / len(close_prices.columns)]
    * len(close_prices.columns)
)
```

```python
transaction_cost = 0.0025

target_weights = train_max_sharpe_weights.copy()

portfolio_value = 1.0
current_weights = target_weights.copy()

portfolio_values = []
portfolio_returns_rebalanced = []
turnovers = []


for date in test_returns.index:

    daily_returns = test_returns.loc[date].values

    # 1. Daily portfolio return before transaction cost
    portfolio_return = np.dot(
        current_weights,
        daily_returns
    )

    # 2. Update portfolio value
    portfolio_value *= (1 + portfolio_return)

    # 3. Update actual weights
    current_weights = (
        current_weights * (1 + daily_returns)
    )

    current_weights = (
        current_weights / np.sum(current_weights)
    )

    # 4. Rebalance at month end
    if date in rebalance_dates:

        turnover = np.sum(
            np.abs(
                current_weights - target_weights
            )
        )

        cost = turnover * transaction_cost

        # Apply transaction cost
        portfolio_value *= (1 - cost)

        # Net return including transaction cost
        portfolio_return = (
            (1 + portfolio_return) * (1 - cost)
            - 1
        )

        # Reset weights
        current_weights = target_weights.copy()

        turnovers.append(turnover)

    # 5. Save results
    portfolio_values.append(portfolio_value)
    portfolio_returns_rebalanced.append(
        portfolio_return
    )


portfolio_values = np.array(portfolio_values)

portfolio_returns_rebalanced = np.array(
    portfolio_returns_rebalanced
)

turnovers = np.array(turnovers)


print("Final Value:", portfolio_values[-1])
print("Total Return:", portfolio_values[-1] - 1)
print("Average Monthly Turnover:", turnovers.mean())
print("Number of Rebalances:", len(turnovers))
```

# Transaction cost sensitivity analysis

```python
transaction_costs = [0.0, 0.001, 0.0025, 0.005]

sensitivity_results = []

for transaction_cost in transaction_costs:
    
    target_weights = train_max_sharpe_weights.copy()
    current_weights = target_weights.copy()
    
    portfolio_value = 1.0
    
    portfolio_values_sensitivity = []
    portfolio_returns_sensitivity = []
    
    for date in test_returns.index:
        
        daily_returns = test_returns.loc[date].values
        
        # Portfolio return before transaction cost
        portfolio_return = np.dot(
            current_weights,
            daily_returns
        )
        
        # Update portfolio value
        portfolio_value *= (1 + portfolio_return)
        
        # Update weights due to price movements
        current_weights = (
            current_weights * (1 + daily_returns)
        )
        
        current_weights = (
            current_weights
            / np.sum(current_weights)
        )
        
        # Rebalance at month-end
        if date in rebalance_dates:
            
            turnover = np.sum(
                np.abs(
                    current_weights
                    - target_weights
                )
            )
            
            cost = turnover * transaction_cost
            
            portfolio_value *= (1 - cost)
            
            # Net return including transaction cost
            portfolio_return = (
                (1 + portfolio_return)
                * (1 - cost)
                - 1
            )
            
            current_weights = target_weights.copy()
        
        portfolio_values_sensitivity.append(
            portfolio_value
        )
        
        portfolio_returns_sensitivity.append(
            portfolio_return
        )
    
    portfolio_values_sensitivity = np.array(
        portfolio_values_sensitivity
    )
    
    portfolio_returns_sensitivity = np.array(
        portfolio_returns_sensitivity
    )
    
    # Performance metrics
    final_value = portfolio_values_sensitivity[-1]
    
    total_return = final_value - 1
    
    annualized_return = (
        final_value
        ** (252 / len(portfolio_returns_sensitivity))
    ) - 1
    
    annualized_volatility = (
        portfolio_returns_sensitivity.std()
        * np.sqrt(252)
    )
    
    sharpe_ratio = (
        annualized_return
        / annualized_volatility
    )
    
    sensitivity_results.append([
        transaction_cost,
        total_return,
        annualized_return,
        annualized_volatility,
        sharpe_ratio
    ])
```

# Walk forward analysis

```python
train_size = 252 * 2
test_size = 252

walk_forward_results = []
```

```python
def optimize_max_sharpe(train_returns):
    
    annualized_returns = train_returns.mean() * 252
    annualized_covariance = train_returns.cov() * 252

    def portfolio_sharpe(weights):
        
        portfolio_return = np.dot(
            weights,
            annualized_returns
        )
        
        portfolio_variance = (
            weights.T
            @ annualized_covariance.values
            @ weights
        )
        
        portfolio_volatility = np.sqrt(
            portfolio_variance
        )
        
        return portfolio_return / portfolio_volatility

    def negative_sharpe(weights):
        return -portfolio_sharpe(weights)

    initial_weights = np.array(
        [1 / len(train_returns.columns)]
        * len(train_returns.columns)
    )

    bounds = [(0, 1)] * len(train_returns.columns)

    constraints = {
        "type": "eq",
        "fun": lambda weights: np.sum(weights) - 1
    }

    result = minimize(
        negative_sharpe,
        initial_weights,
        method="SLSQP",
        bounds=bounds,
        constraints=constraints
    )

    return result.x
```

```python
all_returns = returns.copy()

start = 0

while start + train_size + test_size <= len(all_returns):
    
    train_end = start + train_size
    test_end = train_end + test_size
    
    train_data = all_returns.iloc[start:train_end]
    test_data = all_returns.iloc[train_end:test_end]
    
    # Optimize using training data only
    weights = optimize_max_sharpe(train_data)
    
    # Apply weights to the next unseen test period
    test_portfolio_returns = np.dot(
        test_data,
        weights
    )
    
    cumulative_return = (
        1 + test_portfolio_returns
    ).prod()
    
    total_return = cumulative_return - 1
    
    annualized_return = (
        cumulative_return
        ** (252 / len(test_portfolio_returns))
    ) - 1
    
    annualized_volatility = (
        test_portfolio_returns.std()
        * np.sqrt(252)
    )
    
    sharpe_ratio = (
        annualized_return
        / annualized_volatility
    )
    
    running_max = np.maximum.accumulate(
        (1 + test_portfolio_returns).cumprod()
    )
    
    cumulative_test = (
        1 + test_portfolio_returns
    ).cumprod()
    
    drawdown = (
        cumulative_test / running_max
    ) - 1
    
    max_drawdown = drawdown.min()
    
    walk_forward_results.append([
        test_data.index[0],
        test_data.index[-1],
        total_return,
        annualized_return,
        annualized_volatility,
        sharpe_ratio,
        max_drawdown
    ])
    
    start += test_size
```
